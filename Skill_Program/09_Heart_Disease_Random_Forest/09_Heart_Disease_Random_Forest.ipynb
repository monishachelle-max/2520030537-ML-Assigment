```python
# ============================================================
# PROJECT 9: HEART DISEASE
# RANDOM FOREST ENSEMBLE CLASSIFICATION
#
# Input:
# heart-disease.csv
#
# Output:
# heart_disease_random_forest_results.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.model_selection import train_test_split

from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# PROJECT TITLE
# ============================================================

print("=" * 70)
print("PROJECT 9: HEART DISEASE RANDOM FOREST ENSEMBLE")
print("=" * 70)


# ============================================================
# 2. CHECK CURRENT WORKING DIRECTORY
# ============================================================

print("\nCURRENT WORKING DIRECTORY:")
print(os.getcwd())


# ============================================================
# 3. DEFINE INPUT FILE
# ============================================================

input_file = "heart-disease.csv"


# ============================================================
# 4. CHECK INPUT FILE
# ============================================================

if not os.path.exists(input_file):

    print("\nAVAILABLE CSV FILES:")

    csv_files = [
        file
        for file in os.listdir()
        if file.lower().endswith(".csv")
    ]

    print(csv_files)


    raise FileNotFoundError(

        "\nERROR: The file 'heart-disease(3).csv' was not found.\n\n"
        "Please make sure these two files are in the SAME folder:\n\n"
        "1. 09_Heart_Disease_Random_Forest.ipynb\n"
        "2. heart-disease(3).csv\n"
    )


print("\nINPUT FILE FOUND SUCCESSFULLY!")
print(input_file)


# ============================================================
# 5. LOAD DATASET
# ============================================================

df = pd.read_csv(input_file)


print("\nDATASET LOADED SUCCESSFULLY!")


# ============================================================
# 6. CLEAN COLUMN NAMES
# ============================================================

df.columns = (

    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)

)


print("\nDATASET COLUMNS:")
print(df.columns.tolist())


# ============================================================
# 7. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(
    df.head()
)


# ============================================================
# 8. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")
print(df.shape)


# ============================================================
# 9. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 10. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe()
)


# ============================================================
# 11. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(
    df.isnull().sum()
)


# ============================================================
# 12. DEFINE TARGET COLUMN
# ============================================================

target_column = "target"


# ============================================================
# 13. CHECK TARGET COLUMN
# ============================================================

if target_column not in df.columns:

    raise ValueError(

        "\nERROR: Target column 'target' was not found.\n\n"
        "Available columns are:\n"
        + str(df.columns.tolist())

    )


print("\nTARGET COLUMN FOUND SUCCESSFULLY!")

print(
    target_column
)


# ============================================================
# 14. DEFINE FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=[
        target_column
    ]
)


y = df[
    target_column
].copy()


# ============================================================
# 15. CONVERT TARGET TO NUMERIC
# ============================================================

y = pd.to_numeric(
    y,
    errors="coerce"
)


# ============================================================
# 16. REMOVE MISSING TARGET ROWS
# ============================================================

valid_rows = y.notna()


X = X.loc[
    valid_rows
].copy()


y = y.loc[
    valid_rows
].copy()


# ============================================================
# 17. ENSURE BINARY TARGET
# ============================================================

unique_values = sorted(
    y.unique()
)


print("\nORIGINAL TARGET VALUES:")

print(
    unique_values
)


# If values are not already 0 and 1,
# convert values greater than 0 into Disease

if not set(unique_values).issubset(
    {0, 1}
):

    y = (
        y > 0
    ).astype(int)

else:

    y = y.astype(int)


# ============================================================
# 18. TARGET DISTRIBUTION
# ============================================================

print("\nTARGET DISTRIBUTION:")

print(
    y.value_counts()
)


# ============================================================
# 19. VISUALIZE TARGET DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.countplot(
    x=y
)


plt.title(
    "Heart Disease Class Distribution"
)


plt.xlabel(
    "Heart Disease"
)


plt.ylabel(
    "Count"
)


plt.xticks(

    ticks=[0, 1],

    labels=[
        "No Disease",
        "Disease"
    ]

)


plt.tight_layout()

plt.show()


# ============================================================
# 20. ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(
    X,
    drop_first=False
)


# ============================================================
# 21. CONVERT FEATURES TO NUMERIC
# ============================================================

X = X.apply(
    pd.to_numeric,
    errors="coerce"
)


print("\nFEATURE COLUMNS:")

print(
    X.columns.tolist()
)


print("\nFEATURE DATA:")

display(
    X.head()
)


# ============================================================
# 22. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)


print("\nTRAINING DATA SHAPE:")

print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")

print(
    X_test.shape
)


# ============================================================
# 23. HANDLE MISSING VALUES
#
# IMPORTANT:
# Fit imputer ONLY on training data.
# This prevents data leakage.
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)


X_train_imputed = imputer.fit_transform(
    X_train
)


X_test_imputed = imputer.transform(
    X_test
)


X_train = pd.DataFrame(

    X_train_imputed,

    columns=X_train.columns,

    index=X_train.index

)


X_test = pd.DataFrame(

    X_test_imputed,

    columns=X_test.columns,

    index=X_test.index

)


print(
    "\nMISSING VALUES HANDLED SUCCESSFULLY!"
)


# ============================================================
# 24. CREATE RANDOM FOREST MODEL
# ============================================================

model = RandomForestClassifier(

    n_estimators=300,

    random_state=42,

    n_jobs=-1

)


print(
    "\nRANDOM FOREST MODEL CREATED SUCCESSFULLY!"
)


# ============================================================
# 25. TRAIN RANDOM FOREST MODEL
# ============================================================

model.fit(

    X_train,

    y_train

)


print(
    "\nRANDOM FOREST MODEL TRAINED SUCCESSFULLY!"
)


# ============================================================
# 26. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(
    X_test
)


print(
    "\nPREDICTIONS COMPLETED SUCCESSFULLY!"
)


# ============================================================
# 27. MODEL ACCURACY
# ============================================================

accuracy = accuracy_score(

    y_test,

    y_pred

)


precision = precision_score(

    y_test,

    y_pred,

    zero_division=0

)


recall = recall_score(

    y_test,

    y_pred,

    zero_division=0

)


f1 = f1_score(

    y_test,

    y_pred,

    zero_division=0

)


# ============================================================
# 28. DISPLAY MODEL EVALUATION
# ============================================================

print("\n" + "=" * 50)

print("MODEL EVALUATION")

print("=" * 50)


print("Accuracy :", accuracy)

print("Precision:", precision)

print("Recall   :", recall)

print("F1 Score :", f1)


# ============================================================
# 29. CLASSIFICATION REPORT
# ============================================================

print("\nCLASSIFICATION REPORT:")


print(

    classification_report(

        y_test,

        y_pred,

        labels=[0, 1],

        target_names=[

            "No Disease",

            "Disease"

        ],

        zero_division=0

    )

)


# ============================================================
# 30. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_pred,

    labels=[0, 1]

)


print("\nCONFUSION MATRIX:")

print(
    cm
)


# ============================================================
# 31. CONFUSION MATRIX VISUALIZATION
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    xticklabels=[

        "No Disease",

        "Disease"

    ],

    yticklabels=[

        "No Disease",

        "Disease"

    ]

)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.title(
    "Heart Disease - Random Forest Confusion Matrix"
)


plt.tight_layout()

plt.show()


# ============================================================
# 32. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": model.feature_importances_

    }

)


feature_importance = (

    feature_importance

    .sort_values(

        by="Importance",

        ascending=False

    )

)


print("\nFEATURE IMPORTANCE:")


display(
    feature_importance
)


# ============================================================
# 33. TOP 10 FEATURE IMPORTANCE GRAPH
# ============================================================

top_features = (

    feature_importance

    .head(10)

)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    top_features["Feature"],

    top_features["Importance"]

)


plt.xlabel(
    "Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 10 Random Forest Feature Importances"
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


# ============================================================
# 34. CREATE RESULTS DATASET
# ============================================================

results = pd.DataFrame(

    {

        "Actual": y_test.values,

        "Predicted": y_pred

    }

)


# ============================================================
# 35. ADD ACTUAL LABELS
# ============================================================

results["Actual_Label"] = np.where(

    results["Actual"] == 1,

    "Disease",

    "No Disease"

)


# ============================================================
# 36. ADD PREDICTED LABELS
# ============================================================

results["Predicted_Label"] = np.where(

    results["Predicted"] == 1,

    "Disease",

    "No Disease"

)


print("\nPREDICTION RESULTS:")


display(
    results.head(10)
)


# ============================================================
# 37. SAVE RESULTS DATASET
# ============================================================

output_file = (
    "heart_disease_random_forest_results.csv"
)


results.to_csv(

    output_file,

    index=False

)


print(
    "\nRESULTS FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 38. VERIFY OUTPUT FILE
# ============================================================

print("\n" + "=" * 70)

print("PROJECT 9 COMPLETED SUCCESSFULLY!")

print("Heart Disease — Random Forest Ensemble Classification")

print("=" * 70)


print("\nInput File:")

print(
    input_file
)


print("\nTarget Column:")

print(
    target_column
)


print("\nModel:")

print(
    "Random Forest Classifier"
)


print("\nNumber of Trees:")

print(
    300
)


print("\nOutput File:")

print(
    output_file
)


print("\nFile Created:")

print(
    os.path.exists(
        output_file
    )
)


print("\nExact File Location:")

print(
    os.path.abspath(
        output_file
    )
)


print("\n" + "=" * 70)
```
