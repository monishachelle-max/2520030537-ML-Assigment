```python
# ============================================================
# PROJECT 8: HEART DISEASE
# DECISION TREE CLASSIFICATION
#
# Output:
# heart_disease_decision_tree_results.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier, plot_tree

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 2. SETTINGS
# ============================================================

# Output file name

output_file = "heart_disease_decision_tree_results.csv"


# If automatic target detection fails,
# you can manually write the target column name here.
#
# Example:
# MANUAL_TARGET_COLUMN = "heartdisease"
#
# Leave as None for automatic detection.

MANUAL_TARGET_COLUMN = None


# ============================================================
# 3. CHECK CURRENT WORKING DIRECTORY
# ============================================================

print("=" * 70)
print("CURRENT WORKING DIRECTORY")
print("=" * 70)

print(os.getcwd())


# ============================================================
# 4. FIND CSV FILES
# ============================================================

all_csv_files = glob.glob("*.csv")


# Remove the project output file from possible input files

csv_files = [

    file

    for file in all_csv_files

    if os.path.basename(file).lower()
    != output_file.lower()
]


print("\n" + "=" * 70)
print("CSV FILES AVAILABLE FOR INPUT")
print("=" * 70)

print(csv_files)


# Check whether CSV files exist

if len(csv_files) == 0:

    raise FileNotFoundError(

        "\nNo input CSV file found.\n"
        "Make sure your original Heart Disease dataset CSV file "
        "is in the same folder as this Jupyter Notebook.\n"
        "\nNote: The output file is automatically excluded."
    )


# ============================================================
# 5. SELECT HEART DISEASE DATASET
# ============================================================

input_file = None


# Common Heart Disease dataset file names

possible_files = [

    "heart.csv",

    "heart_disease.csv",

    "Heart_Disease.csv",

    "Heart_Disease_Prediction.csv",

    "heart_disease_uci.csv",

    "heart_disease_data.csv",

    "heart_failure_clinical_records_dataset.csv",

    "cardio_train.csv",

    "framingham.csv"
]


# Try exact file names first

for file in possible_files:

    if file in csv_files:

        input_file = file

        break


# If exact file is not found,
# prefer files containing heart-related words

if input_file is None:

    keywords = [

        "heart",
        "cardio",
        "framingham",
        "disease"
    ]


    for file in csv_files:

        file_name = file.lower()


        if any(
            keyword in file_name
            for keyword in keywords
        ):

            input_file = file

            break


# If still not found,
# use the first remaining CSV file

if input_file is None:

    input_file = csv_files[0]


print("\n" + "=" * 70)
print("SELECTED INPUT FILE")
print("=" * 70)

print(input_file)


# ============================================================
# 6. LOAD DATASET
# ============================================================

df = pd.read_csv(input_file)


print("\nHeart Disease Dataset Loaded Successfully!")


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

df.columns = (

    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
    .str.replace("/", "_", regex=False)
)


print("\n" + "=" * 70)
print("CLEANED COLUMN NAMES")
print("=" * 70)

print(df.columns.tolist())


# ============================================================
# 8. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(df.head())


# ============================================================
# 9. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")

print(df.shape)


# ============================================================
# 10. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 11. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe(include="all")
)


# ============================================================
# 12. REPLACE COMMON MISSING VALUE SYMBOLS
# ============================================================

df = df.replace(

    [
        "?",
        " ?",
        "",
        " ",
        "NA",
        "N/A",
        "null",
        "NULL",
        "None",
        "none"
    ],

    np.nan
)


print("\nMissing value symbols replaced.")


# ============================================================
# 13. CONVERT NUMERIC-LIKE COLUMNS
# ============================================================

for column in df.columns:

    converted_column = pd.to_numeric(
        df[column],
        errors="coerce"
    )


    original_non_null = (
        df[column]
        .notna()
        .sum()
    )


    converted_non_null = (
        converted_column
        .notna()
        .sum()
    )


    # Convert column only when all non-null values
    # can successfully be converted to numeric

    if (
        original_non_null > 0
        and converted_non_null == original_non_null
    ):

        df[column] = converted_column


# ============================================================
# 14. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

print(
    df.isnull().sum()
)


# ============================================================
# 15. DISPLAY ALL AVAILABLE COLUMNS
# ============================================================

print("\n" + "=" * 70)
print("AVAILABLE COLUMNS")
print("=" * 70)


for i, column in enumerate(df.columns):

    print(f"{i + 1}. {column}")


# ============================================================
# 16. IDENTIFY TARGET COLUMN
# ============================================================

target_column = None


# ------------------------------------------------------------
# USE MANUAL TARGET COLUMN IF PROVIDED
# ------------------------------------------------------------

if MANUAL_TARGET_COLUMN is not None:

    manual_column = (

        MANUAL_TARGET_COLUMN
        .strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
    )


    if manual_column in df.columns:

        target_column = manual_column


    else:

        raise ValueError(

            f"\nManual target column "
            f"'{MANUAL_TARGET_COLUMN}' "
            f"was not found.\n\n"

            f"Available columns:\n"
            f"{df.columns.tolist()}"
        )


# ------------------------------------------------------------
# COMMON HEART DISEASE TARGET COLUMN NAMES
# ------------------------------------------------------------

if target_column is None:

    possible_target_columns = [

        # Standard ML datasets

        "target",
        "output",
        "label",
        "class",

        # Heart Disease datasets

        "heartdisease",
        "heart_disease",

        "heartdiseasepresent",
        "heart_disease_present",

        "heartdiseaseorattack",
        "heart_disease_or_attack",

        "heartdiseaseprediction",
        "heart_disease_prediction",

        "heartdiseasepresence",
        "heart_disease_presence",

        # UCI dataset

        "num",

        # Diagnosis

        "diagnosis",

        "disease",

        "condition",

        "outcome",

        "result",

        "status",

        # Heart failure dataset

        "death_event",

        # Framingham dataset

        "tenyearchd",

        "ten_year_chd",

        # Cardio dataset

        "cardio",

        # Other common labels

        "presence",

        "present"
    ]


    # Exact match

    for column in possible_target_columns:

        if column in df.columns:

            target_column = column

            break


# ------------------------------------------------------------
# NORMALIZED COLUMN MATCH
# ------------------------------------------------------------

if target_column is None:

    normalized_targets = [

        "target",
        "output",
        "label",
        "class",

        "heartdisease",
        "heartdiseasepresent",

        "heartdiseaseorattack",

        "heartdiseaseprediction",

        "heartdiseasepresence",

        "num",

        "diagnosis",

        "disease",

        "condition",

        "outcome",

        "result",

        "status",

        "deathevent",

        "tenyearchd",

        "cardio"
    ]


    for column in df.columns:

        normalized_column = (

            column
            .lower()
            .replace("_", "")
            .replace(" ", "")
            .replace("-", "")
        )


        if normalized_column in normalized_targets:

            target_column = column

            break


# ------------------------------------------------------------
# KEYWORD DETECTION
# ------------------------------------------------------------

if target_column is None:

    target_keywords = [

        "target",
        "heartdisease",
        "diagnosis",
        "disease",
        "outcome",
        "condition",
        "death_event",
        "death",
        "chd",
        "cardio"
    ]


    for column in df.columns:

        normalized_column = (

            column
            .lower()
            .replace("_", "")
        )


        normalized_keywords = [

            keyword
            .replace("_", "")
            for keyword in target_keywords
        ]


        if normalized_column in normalized_keywords:

            target_column = column

            break


# ============================================================
# TARGET COLUMN NOT FOUND
# ============================================================

if target_column is None:

    print("\n" + "=" * 70)
    print("TARGET COLUMN COULD NOT BE DETECTED")
    print("=" * 70)


    print("\nAVAILABLE COLUMNS:")


    for i, column in enumerate(df.columns):

        print(f"{i + 1}. {column}")


    raise ValueError(

        "\nTarget column could not be detected automatically.\n"
        "\nSet MANUAL_TARGET_COLUMN at the top of the notebook.\n"
        "\nExample:\n"
        'MANUAL_TARGET_COLUMN = "your_target_column_name"'
    )


# ============================================================
# 17. DISPLAY TARGET COLUMN
# ============================================================

print("\n" + "=" * 70)
print("TARGET COLUMN FOUND SUCCESSFULLY")
print("=" * 70)

print("Target Column:", target_column)


# ============================================================
# 18. PREPARE TARGET VARIABLE
# ============================================================

y = df[target_column].copy()


# Remove rows with missing target values

valid_target = y.notna()


df = df.loc[
    valid_target
].copy()


y = y.loc[
    valid_target
].copy()


# ============================================================
# 19. CONVERT TARGET TO BINARY
# ============================================================

y_numeric = pd.to_numeric(
    y,
    errors="coerce"
)


# ------------------------------------------------------------
# CASE 1: NUMERIC TARGET
# ------------------------------------------------------------

if y_numeric.notna().sum() == len(y):

    y = y_numeric


    unique_values = sorted(
        y.unique()
    )


    print("\nORIGINAL NUMERIC TARGET VALUES:")

    print(unique_values)


    # Already binary

    if set(unique_values).issubset(
        {0, 1}
    ):

        y = y.astype(int)


    # Multi-class:
    # 0 = No Disease
    # > 0 = Disease

    else:

        y = (
            y > 0
        ).astype(int)


# ------------------------------------------------------------
# CASE 2: TEXT TARGET
# ------------------------------------------------------------

else:

    y_text = (

        y.astype(str)
        .str.strip()
        .str.lower()
    )


    print("\nORIGINAL TEXT TARGET VALUES:")

    print(
        y_text.value_counts()
    )


    unique_text_values = list(
        y_text.unique()
    )


    disease_values = [

        "1",

        "yes",

        "y",

        "true",

        "positive",

        "disease",

        "heart disease",

        "presence",

        "present",

        "abnormal",

        "disease_present"
    ]


    no_disease_values = [

        "0",

        "no",

        "n",

        "false",

        "negative",

        "no disease",

        "absence",

        "absent",

        "normal",

        "healthy"
    ]


    # If common labels are detected

    if all(

        value in (
            disease_values
            + no_disease_values
        )

        for value in unique_text_values
    ):

        y = np.where(

            y_text.isin(
                disease_values
            ),

            1,

            0
        )


    # If exactly two unknown text classes exist

    elif len(unique_text_values) == 2:

        unique_text_values = sorted(
            unique_text_values
        )


        print("\nBinary text target detected.")

        print(
            "Class Mapping:"
        )


        print(
            f"{unique_text_values[0]} = 0"
        )


        print(
            f"{unique_text_values[1]} = 1"
        )


        y = pd.Series(

            np.where(

                y_text
                == unique_text_values[1],

                1,

                0
            ),

            index=df.index
        )


    else:

        raise ValueError(

            "\nText target could not be safely "
            "converted to binary.\n"

            f"Target values are:\n"
            f"{unique_text_values}"
        )


y = pd.Series(
    y,
    index=df.index
).astype(int)


# ============================================================
# 20. TARGET DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("FINAL TARGET DISTRIBUTION")
print("=" * 70)

print(
    y.value_counts()
)


# Verify two classes exist

if y.nunique() != 2:

    raise ValueError(

        "\nThe target must contain exactly "
        "two classes after processing."
    )


# ============================================================
# 21. VISUALIZE TARGET DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.countplot(
    x=y
)


plt.title(
    "Heart Disease Class Distribution"
)


plt.xlabel(
    "Heart Disease"
)


plt.ylabel(
    "Count"
)


plt.xticks(

    ticks=[0, 1],

    labels=[
        "No Disease",
        "Disease"
    ]
)


plt.tight_layout()

plt.show()


# ============================================================
# 22. DEFINE FEATURES
# ============================================================

X = df.drop(
    columns=[
        target_column
    ]
)


print("\n" + "=" * 70)
print("FEATURE COLUMNS")
print("=" * 70)

print(
    X.columns.tolist()
)


# ============================================================
# 23. ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(
    X,
    drop_first=True
)


# Convert boolean values to numeric

X = X.astype(float)


print("\nFEATURE DATA AFTER ENCODING:")

display(
    X.head()
)


# ============================================================
# 24. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\nTRAINING DATA SHAPE:")

print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")

print(
    X_test.shape
)


# ============================================================
# 25. HANDLE MISSING VALUES
# ============================================================

# Fit imputer only on training data

imputer = SimpleImputer(
    strategy="median"
)


X_train_imputed = imputer.fit_transform(
    X_train
)


X_test_imputed = imputer.transform(
    X_test
)


X_train = pd.DataFrame(

    X_train_imputed,

    columns=X.columns,

    index=X_train.index
)


X_test = pd.DataFrame(

    X_test_imputed,

    columns=X.columns,

    index=X_test.index
)


print(
    "\nMissing values handled successfully!"
)


# ============================================================
# 26. CREATE DECISION TREE MODEL
# ============================================================

model = DecisionTreeClassifier(

    criterion="gini",

    max_depth=5,

    random_state=42
)


print(
    "\nDecision Tree Model Created Successfully!"
)


# ============================================================
# 27. TRAIN MODEL
# ============================================================

model.fit(

    X_train,

    y_train
)


print(
    "\nDecision Tree Model Trained Successfully!"
)


# ============================================================
# 28. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(
    X_test
)


print(
    "\nPredictions Completed Successfully!"
)


# ============================================================
# 29. MODEL EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)


precision = precision_score(

    y_test,

    y_pred,

    zero_division=0
)


recall = recall_score(

    y_test,

    y_pred,

    zero_division=0
)


f1 = f1_score(

    y_test,

    y_pred,

    zero_division=0
)


print("\n" + "=" * 50)
print("MODEL EVALUATION")
print("=" * 50)

print("Accuracy :", accuracy)

print("Precision:", precision)

print("Recall   :", recall)

print("F1 Score :", f1)


# ============================================================
# 30. CLASSIFICATION REPORT
# ============================================================

print(
    "\nCLASSIFICATION REPORT:"
)


print(

    classification_report(

        y_test,

        y_pred,

        labels=[0, 1],

        target_names=[

            "No Disease",

            "Disease"
        ],

        zero_division=0
    )
)


# ============================================================
# 31. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_pred,

    labels=[0, 1]
)


print(
    "\nCONFUSION MATRIX:"
)


print(cm)


# ============================================================
# 32. CONFUSION MATRIX VISUALIZATION
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    xticklabels=[

        "No Disease",

        "Disease"
    ],

    yticklabels=[

        "No Disease",

        "Disease"
    ]
)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.title(
    "Heart Disease - Decision Tree Confusion Matrix"
)


plt.tight_layout()

plt.show()


# ============================================================
# 33. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": model.feature_importances_
    }
)


feature_importance = (

    feature_importance

    .sort_values(

        by="Importance",

        ascending=False
    )
)


print(
    "\nFEATURE IMPORTANCE:"
)


display(
    feature_importance
)


# ============================================================
# 34. FEATURE IMPORTANCE GRAPH
# ============================================================

top_features = (

    feature_importance

    .head(10)
)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    top_features["Feature"],

    top_features["Importance"]
)


plt.xlabel(
    "Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 10 Decision Tree Feature Importances"
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


# ============================================================
# 35. VISUALIZE DECISION TREE
# ============================================================

plt.figure(
    figsize=(20, 10)
)


plot_tree(

    model,

    feature_names=X.columns,

    class_names=[

        "No Disease",

        "Disease"
    ],

    filled=True,

    rounded=True,

    fontsize=8
)


plt.title(
    "Heart Disease Decision Tree"
)


plt.tight_layout()

plt.show()


# ============================================================
# 36. CREATE RESULTS DATASET
# ============================================================

results = pd.DataFrame(

    {

        "Actual": y_test.values,

        "Predicted": y_pred
    }
)


results["Actual_Label"] = np.where(

    results["Actual"] == 1,

    "Disease",

    "No Disease"
)


results["Predicted_Label"] = np.where(

    results["Predicted"] == 1,

    "Disease",

    "No Disease"
)


print(
    "\nPREDICTION RESULTS:"
)


display(
    results.head(10)
)


# ============================================================
# 37. SAVE RESULTS DATASET
# ============================================================

results.to_csv(

    output_file,

    index=False
)


# ============================================================
# 38. VERIFY OUTPUT FILE
# ============================================================

print("\n" + "=" * 70)
print("PROJECT 8 COMPLETED SUCCESSFULLY!")
print("Heart Disease — Decision Tree Classification")
print("=" * 70)


print("\nInput File:")

print(input_file)


print("\nTarget Column:")

print(target_column)


print("\nModel:")

print(
    "Decision Tree Classifier"
)


print("\nCriterion:")

print(
    "Gini Impurity"
)


print("\nMaximum Depth:")

print(5)


print("\nOutput File:")

print(output_file)


print("\nFile Created:")

print(
    os.path.exists(
        output_file
    )
)


print("\nExact File Location:")

print(
    os.path.abspath(
        output_file
    )
)


print("\n" + "=" * 70)
```
