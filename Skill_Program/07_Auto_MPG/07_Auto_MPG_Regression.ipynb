```python
# ============================================================
# PROJECT 7: AUTO MPG
# LINEAR REGRESSION, FEATURE SCALING, AND ENCODING
#
# Notebook:
# 07_Auto_MPG_Regression.ipynb
#
# Input:
# auto-mpg.csv
#
# Output:
# auto_mpg_regression_results.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. LOAD AUTO MPG DATASET
# ============================================================

df = pd.read_csv("auto-mpg.csv")

print("Auto MPG Dataset Loaded Successfully!")


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(df.head())


# ============================================================
# 4. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")

print(df.shape)


# ============================================================
# 5. COLUMN NAMES
# ============================================================

print("\nCOLUMN NAMES:")

print(df.columns.tolist())


# ============================================================
# 6. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)


print("\nCLEANED COLUMN NAMES:")

print(df.columns.tolist())


# ============================================================
# 7. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 8. REPLACE QUESTION MARKS WITH MISSING VALUES
# ============================================================

df = df.replace("?", np.nan)

print("\nQuestion marks replaced with missing values.")


# ============================================================
# 9. CONVERT HORSEPOWER TO NUMERIC
# ============================================================

df["horsepower"] = pd.to_numeric(
    df["horsepower"],
    errors="coerce"
)

print("\nHorsepower converted to numeric successfully.")


# ============================================================
# 10. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(df.isnull().sum())


# ============================================================
# 11. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe(
        include="all"
    )
)


# ============================================================
# 12. MPG DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(8, 5)
)

sns.histplot(
    data=df,
    x="mpg",
    kde=True,
    bins=30
)

plt.title(
    "MPG Distribution"
)

plt.xlabel(
    "MPG"
)

plt.ylabel(
    "Count"
)

plt.tight_layout()

plt.show()


# ============================================================
# 13. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=["number"]
)


plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Auto MPG Correlation Heatmap"
)

plt.tight_layout()

plt.show()


# ============================================================
# 14. DEFINE TARGET VARIABLE
# ============================================================

y = df["mpg"]


print("\nTARGET VARIABLE:")

print("mpg")


# ============================================================
# 15. DEFINE FEATURE VARIABLES
# ============================================================

# car_name contains mostly unique car names,
# so it is dropped.

X = df.drop(
    columns=[
        "mpg",
        "car_name"
    ]
)


print("\nFEATURE COLUMNS:")

print(
    X.columns.tolist()
)


# ============================================================
# 16. DEFINE NUMERICAL FEATURES
# ============================================================

numerical_columns = [
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year"
]


# ============================================================
# 17. DEFINE CATEGORICAL FEATURES
# ============================================================

# origin represents country/region category.
# Therefore it is explicitly treated as categorical
# and encoded using OneHotEncoder.

categorical_columns = [
    "origin"
]


print("\nNUMERICAL COLUMNS:")

print(
    numerical_columns
)


print("\nCATEGORICAL COLUMNS:")

print(
    categorical_columns
)


# ============================================================
# 18. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("\nTRAINING DATA SHAPE:")

print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")

print(
    X_test.shape
)


# ============================================================
# 19. NUMERICAL PREPROCESSING
# ============================================================

# Handle missing numerical values
# and scale numerical features.

numerical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ============================================================
# 20. CATEGORICAL PREPROCESSING
# ============================================================

# Handle missing categorical values
# and encode categorical features.

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


# ============================================================
# 21. CREATE PREPROCESSING PIPELINE
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical_features",
            numerical_transformer,
            numerical_columns
        ),

        (
            "categorical_features",
            categorical_transformer,
            categorical_columns
        )
    ]
)


print(
    "\nFeature Scaling and Encoding Pipeline Created!"
)


# ============================================================
# 22. CREATE LINEAR REGRESSION PIPELINE
# ============================================================

model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "linear_regression",
            LinearRegression()
        )
    ]
)


print(
    "\nLinear Regression Model Created!"
)


# ============================================================
# 23. TRAIN THE MODEL
# ============================================================

model.fit(
    X_train,
    y_train
)


print(
    "\nModel Trained Successfully!"
)


# ============================================================
# 24. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(
    X_test
)


print(
    "\nPredictions Completed Successfully!"
)


# ============================================================
# 25. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)


mse = mean_squared_error(
    y_test,
    y_pred
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(
    y_test,
    y_pred
)


print("\n==============================================")
print("MODEL EVALUATION")
print("==============================================")

print("MAE      :", mae)

print("MSE      :", mse)

print("RMSE     :", rmse)

print("R2 Score :", r2)


# ============================================================
# 26. CREATE RESULTS DATASET
# ============================================================

results = pd.DataFrame(
    {
        "Actual_MPG": y_test.values,
        "Predicted_MPG": y_pred,
        "Prediction_Error": (
            y_test.values - y_pred
        )
    }
)


print(
    "\nACTUAL VS PREDICTED RESULTS:"
)


display(
    results.head(10)
)


# ============================================================
# 27. ACTUAL VS PREDICTED GRAPH
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.scatter(
    y_test,
    y_pred
)


minimum_value = min(
    y_test.min(),
    y_pred.min()
)


maximum_value = max(
    y_test.max(),
    y_pred.max()
)


plt.plot(
    [
        minimum_value,
        maximum_value
    ],
    [
        minimum_value,
        maximum_value
    ]
)


plt.xlabel(
    "Actual MPG"
)


plt.ylabel(
    "Predicted MPG"
)


plt.title(
    "Auto MPG - Actual vs Predicted"
)


plt.tight_layout()

plt.show()


# ============================================================
# 28. RESIDUAL PLOT
# ============================================================

residuals = (
    y_test.values - y_pred
)


plt.figure(
    figsize=(8, 5)
)


plt.scatter(
    y_pred,
    residuals
)


plt.axhline(
    y=0
)


plt.xlabel(
    "Predicted MPG"
)


plt.ylabel(
    "Residual Error"
)


plt.title(
    "Auto MPG - Residual Plot"
)


plt.tight_layout()

plt.show()


# ============================================================
# 29. SAVE RESULTS DATASET
# ============================================================

output_file = (
    "auto_mpg_regression_results.csv"
)


results.to_csv(
    output_file,
    index=False
)


# ============================================================
# 30. VERIFY OUTPUT FILE
# ============================================================

print("\n==============================================")
print("PROJECT 7 COMPLETED SUCCESSFULLY!")
print("Auto MPG — Linear Regression, Feature Scaling, and Encoding")
print("==============================================")

print("\nInput File:")
print("auto-mpg.csv")

print("\nTarget:")
print("mpg")

print("\nModel:")
print("Linear Regression")

print("\nFeature Scaling:")
print("StandardScaler")

print("\nCategorical Encoding:")
print("OneHotEncoder")

print("\nEncoded Feature:")
print("origin")

print("\nOutput File:")
print(output_file)

print("\nFile Created:")

print(
    os.path.exists(
        output_file
    )
)


print("\nExact File Location:")

print(
    os.path.abspath(
        output_file
    )
)


print("\n==============================================")
```
