```python
# ============================================================
# PROJECT 1: TITANIC SURVIVAL
# EDA AND ML LIFECYCLE MAPPING
# File: 01_Titanic_EDA_ML_Lifecycle.ipynb
# Input: titanic.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("titanic.csv")

print("Dataset loaded successfully!")


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(df.head())


# ============================================================
# 4. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")

print(df.shape)


# ============================================================
# 5. COLUMN NAMES
# ============================================================

print("\nCOLUMN NAMES:")

print(df.columns.tolist())


# ============================================================
# 6. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 7. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(df.describe(include="all"))


# ============================================================
# 8. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(df.isnull().sum())


# ============================================================
# 9. SURVIVAL DISTRIBUTION
# ============================================================

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="survived"
)

plt.title("Titanic Survival Distribution")

plt.xlabel("Survived (0 = No, 1 = Yes)")

plt.ylabel("Passenger Count")

plt.tight_layout()

plt.show()


# ============================================================
# 10. SURVIVAL BY GENDER
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sex",
    hue="survived"
)

plt.title("Survival by Gender")

plt.xlabel("Gender")

plt.ylabel("Passenger Count")

plt.tight_layout()

plt.show()


# ============================================================
# 11. SURVIVAL BY PASSENGER CLASS
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="pclass",
    hue="survived"
)

plt.title("Survival by Passenger Class")

plt.xlabel("Passenger Class")

plt.ylabel("Passenger Count")

plt.tight_layout()

plt.show()


# ============================================================
# 12. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    kde=True,
    bins=30
)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Passenger Count")

plt.tight_layout()

plt.show()


# ============================================================
# 13. FARE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="fare",
    kde=True,
    bins=30
)

plt.title("Fare Distribution")

plt.xlabel("Fare")

plt.ylabel("Passenger Count")

plt.tight_layout()

plt.show()


# ============================================================
# 14. SURVIVAL BY EMBARKATION PORT
# ============================================================

if "embarked" in df.columns:

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df,
        x="embarked",
        hue="survived"
    )

    plt.title("Survival by Embarkation Port")

    plt.xlabel("Embarkation Port")

    plt.ylabel("Passenger Count")

    plt.tight_layout()

    plt.show()


# ============================================================
# 15. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include="number"
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f"
)

plt.title("Titanic Dataset Correlation Heatmap")

plt.tight_layout()

plt.show()


# ============================================================
# 16. ML LIFECYCLE MAPPING
# ============================================================

print("\n================ ML LIFECYCLE MAPPING ================\n")

print("1. PROBLEM DEFINITION")
print("   Predict whether a Titanic passenger survived or not.")

print("\n2. DATA COLLECTION")
print("   Dataset used: titanic.csv")

print("\n3. DATA UNDERSTANDING")
print("   Checked dataset shape, columns, data types and statistics.")

print("\n4. DATA PREPROCESSING")
print("   Missing values and categorical features need preprocessing.")

print("\n5. EXPLORATORY DATA ANALYSIS")
print("   Analysed survival using gender, class, age, fare and embarkation.")

print("\n6. FEATURE ENGINEERING")
print("   Relevant features can be prepared for machine learning.")

print("\n7. MODEL BUILDING")
print("   A classification algorithm can be trained to predict survival.")

print("\n8. MODEL EVALUATION")
print("   Model performance can be evaluated using accuracy and other metrics.")

print("\n9. DEPLOYMENT")
print("   The trained model can be used to predict survival for new passengers.")


# ============================================================
# 17. PROJECT COMPLETION MESSAGE
# ============================================================

print("\n==============================================")

print("PROJECT 1 COMPLETED SUCCESSFULLY!")

print("Titanic Survival — EDA and ML Lifecycle Mapping")

print("==============================================")
```
