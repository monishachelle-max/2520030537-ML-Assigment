```python
# ============================================================
# PROJECT 11: WINE QUALITY
# DECISION TREE AND RANDOM FOREST CLASSIFICATION
#
# Input:
# winequality-red.csv
#
# Outputs:
# wine_quality_predictions.csv
# wine_quality_feature_importance.csv
# wine_quality_model_comparison.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.impute import SimpleImputer

from sklearn.tree import (
    DecisionTreeClassifier,
    plot_tree
)

from sklearn.ensemble import (
    RandomForestClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 2. PROJECT INFORMATION
# ============================================================

print("=" * 70)
print("PROJECT 11: WINE QUALITY")
print("DECISION TREE AND RANDOM FOREST CLASSIFICATION")
print("=" * 70)


# ============================================================
# 3. CHECK CURRENT WORKING DIRECTORY
# ============================================================

print("\nCURRENT WORKING DIRECTORY:")
print(os.getcwd())


# ============================================================
# 4. DEFINE INPUT FILE
# ============================================================

input_file = "winequality-red.csv"


# ============================================================
# 5. CHECK INPUT FILE
# ============================================================

if not os.path.exists(input_file):

    raise FileNotFoundError(
        "\nERROR: winequality-red.csv was not found.\n\n"
        "Please make sure your folder contains:\n"
        "1. 11_Wine_Quality_Decision_Tree_Random_Forest.ipynb\n"
        "2. winequality-red.csv"
    )


print("\nINPUT FILE FOUND:")
print(input_file)


# ============================================================
# 6. LOAD DATASET
# ============================================================

# Automatically detects comma or semicolon separator

df = pd.read_csv(
    input_file,
    sep=None,
    engine="python"
)


print("\nWINE QUALITY DATASET LOADED SUCCESSFULLY!")


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)


print("\nDATASET COLUMNS:")
print(df.columns.tolist())


# ============================================================
# 8. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(
    df.head()
)


# ============================================================
# 9. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")
print(df.shape)


# ============================================================
# 10. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 11. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe()
)


# ============================================================
# 12. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(
    df.isnull().sum()
)


# ============================================================
# 13. CHECK TARGET COLUMN
# ============================================================

target_column = "quality"


if target_column not in df.columns:

    raise ValueError(
        "\nERROR: 'quality' column not found.\n"
        "Available columns:\n"
        + str(df.columns.tolist())
    )


print("\nTARGET COLUMN FOUND:")
print(target_column)


# ============================================================
# 14. ORIGINAL QUALITY DISTRIBUTION
# ============================================================

print("\nORIGINAL WINE QUALITY DISTRIBUTION:")

print(
    df[target_column]
    .value_counts()
    .sort_index()
)


# ============================================================
# 15. VISUALIZE ORIGINAL QUALITY DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(8, 5)
)


sns.countplot(
    x=df[target_column]
)


plt.title(
    "Original Wine Quality Distribution"
)


plt.xlabel(
    "Wine Quality Score"
)


plt.ylabel(
    "Count"
)


plt.tight_layout()

plt.show()


# ============================================================
# 16. CONVERT QUALITY INTO BINARY CLASSES
# ============================================================

# Quality 0 to 6 = Low Quality
# Quality 7 to 10 = High Quality

y = (
    df[target_column] >= 7
).astype(int)


print("\nBINARY TARGET CREATED SUCCESSFULLY!")

print("\n0 = Low Quality")
print("1 = High Quality")


# ============================================================
# 17. BINARY TARGET DISTRIBUTION
# ============================================================

print("\nBINARY QUALITY DISTRIBUTION:")

print(
    y.value_counts()
)


# ============================================================
# 18. VISUALIZE BINARY TARGET
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.countplot(
    x=y
)


plt.title(
    "Wine Quality Class Distribution"
)


plt.xlabel(
    "Wine Quality Class"
)


plt.ylabel(
    "Count"
)


plt.xticks(
    ticks=[0, 1],
    labels=[
        "Low Quality",
        "High Quality"
    ]
)


plt.tight_layout()

plt.show()


# ============================================================
# 19. DEFINE FEATURES
# ============================================================

X = df.drop(
    columns=[
        target_column
    ]
)


print("\nFEATURE COLUMNS:")

print(
    X.columns.tolist()
)


# ============================================================
# 20. CONVERT FEATURES TO NUMERIC
# ============================================================

for column in X.columns:

    X[column] = pd.to_numeric(
        X[column],
        errors="coerce"
    )


# ============================================================
# 21. HANDLE MISSING VALUES
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)


X_imputed = imputer.fit_transform(
    X
)


X = pd.DataFrame(
    X_imputed,
    columns=X.columns,
    index=X.index
)


print("\nMISSING VALUES HANDLED SUCCESSFULLY!")


# ============================================================
# 22. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\nTRAINING DATA SHAPE:")
print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")
print(
    X_test.shape
)


# ============================================================
# 23. CREATE DECISION TREE MODEL
# ============================================================

decision_tree_model = DecisionTreeClassifier(

    criterion="gini",

    max_depth=5,

    random_state=42,

    class_weight="balanced"
)


print(
    "\nDECISION TREE MODEL CREATED SUCCESSFULLY!"
)


# ============================================================
# 24. TRAIN DECISION TREE MODEL
# ============================================================

decision_tree_model.fit(

    X_train,

    y_train
)


print(
    "\nDECISION TREE MODEL TRAINED SUCCESSFULLY!"
)


# ============================================================
# 25. DECISION TREE PREDICTIONS
# ============================================================

dt_pred = decision_tree_model.predict(
    X_test
)


print(
    "\nDECISION TREE PREDICTIONS COMPLETED!"
)


# ============================================================
# 26. DECISION TREE EVALUATION
# ============================================================

dt_accuracy = accuracy_score(
    y_test,
    dt_pred
)


dt_precision = precision_score(
    y_test,
    dt_pred,
    zero_division=0
)


dt_recall = recall_score(
    y_test,
    dt_pred,
    zero_division=0
)


dt_f1 = f1_score(
    y_test,
    dt_pred,
    zero_division=0
)


print("\n" + "=" * 60)
print("DECISION TREE MODEL EVALUATION")
print("=" * 60)

print("Accuracy :", dt_accuracy)
print("Precision:", dt_precision)
print("Recall   :", dt_recall)
print("F1 Score :", dt_f1)


# ============================================================
# 27. DECISION TREE CLASSIFICATION REPORT
# ============================================================

print(
    "\nDECISION TREE CLASSIFICATION REPORT:"
)


print(
    classification_report(

        y_test,

        dt_pred,

        labels=[0, 1],

        target_names=[
            "Low Quality",
            "High Quality"
        ],

        zero_division=0
    )
)


# ============================================================
# 28. DECISION TREE CONFUSION MATRIX
# ============================================================

dt_cm = confusion_matrix(

    y_test,

    dt_pred,

    labels=[0, 1]
)


print(
    "\nDECISION TREE CONFUSION MATRIX:"
)


print(
    dt_cm
)


# ============================================================
# 29. DECISION TREE CONFUSION MATRIX GRAPH
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    dt_cm,

    annot=True,

    fmt="d",

    xticklabels=[
        "Low Quality",
        "High Quality"
    ],

    yticklabels=[
        "Low Quality",
        "High Quality"
    ]
)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.title(
    "Decision Tree Confusion Matrix"
)


plt.tight_layout()

plt.show()


# ============================================================
# 30. CREATE RANDOM FOREST MODEL
# ============================================================

random_forest_model = RandomForestClassifier(

    n_estimators=200,

    max_depth=10,

    random_state=42,

    class_weight="balanced",

    n_jobs=-1
)


print(
    "\nRANDOM FOREST MODEL CREATED SUCCESSFULLY!"
)


# ============================================================
# 31. TRAIN RANDOM FOREST MODEL
# ============================================================

random_forest_model.fit(

    X_train,

    y_train
)


print(
    "\nRANDOM FOREST MODEL TRAINED SUCCESSFULLY!"
)


# ============================================================
# 32. RANDOM FOREST PREDICTIONS
# ============================================================

rf_pred = random_forest_model.predict(
    X_test
)


print(
    "\nRANDOM FOREST PREDICTIONS COMPLETED!"
)


# ============================================================
# 33. RANDOM FOREST EVALUATION
# ============================================================

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)


rf_precision = precision_score(
    y_test,
    rf_pred,
    zero_division=0
)


rf_recall = recall_score(
    y_test,
    rf_pred,
    zero_division=0
)


rf_f1 = f1_score(
    y_test,
    rf_pred,
    zero_division=0
)


print("\n" + "=" * 60)
print("RANDOM FOREST MODEL EVALUATION")
print("=" * 60)

print("Accuracy :", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1 Score :", rf_f1)


# ============================================================
# 34. RANDOM FOREST CLASSIFICATION REPORT
# ============================================================

print(
    "\nRANDOM FOREST CLASSIFICATION REPORT:"
)


print(
    classification_report(

        y_test,

        rf_pred,

        labels=[0, 1],

        target_names=[
            "Low Quality",
            "High Quality"
        ],

        zero_division=0
    )
)


# ============================================================
# 35. RANDOM FOREST CONFUSION MATRIX
# ============================================================

rf_cm = confusion_matrix(

    y_test,

    rf_pred,

    labels=[0, 1]
)


print(
    "\nRANDOM FOREST CONFUSION MATRIX:"
)


print(
    rf_cm
)


# ============================================================
# 36. RANDOM FOREST CONFUSION MATRIX GRAPH
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    rf_cm,

    annot=True,

    fmt="d",

    xticklabels=[
        "Low Quality",
        "High Quality"
    ],

    yticklabels=[
        "Low Quality",
        "High Quality"
    ]
)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.title(
    "Random Forest Confusion Matrix"
)


plt.tight_layout()

plt.show()


# ============================================================
# 37. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame(

    {

        "Model": [

            "Decision Tree",

            "Random Forest"
        ],

        "Accuracy": [

            dt_accuracy,

            rf_accuracy
        ],

        "Precision": [

            dt_precision,

            rf_precision
        ],

        "Recall": [

            dt_recall,

            rf_recall
        ],

        "F1 Score": [

            dt_f1,

            rf_f1
        ]
    }
)


print("\n" + "=" * 60)
print("MODEL COMPARISON")
print("=" * 60)


display(
    comparison
)


# ============================================================
# 38. MODEL COMPARISON GRAPH
# ============================================================

comparison_plot = comparison.set_index(
    "Model"
)


comparison_plot.plot(

    kind="bar",

    figsize=(10, 6)
)


plt.title(
    "Decision Tree vs Random Forest Performance"
)


plt.ylabel(
    "Score"
)


plt.ylim(
    0,
    1
)


plt.xticks(
    rotation=0
)


plt.tight_layout()

plt.show()


# ============================================================
# 39. DECISION TREE FEATURE IMPORTANCE
# ============================================================

dt_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": decision_tree_model.feature_importances_
    }
)


dt_importance = dt_importance.sort_values(

    by="Importance",

    ascending=False
)


print(
    "\nDECISION TREE FEATURE IMPORTANCE:"
)


display(
    dt_importance
)


# ============================================================
# 40. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

rf_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": random_forest_model.feature_importances_
    }
)


rf_importance = rf_importance.sort_values(

    by="Importance",

    ascending=False
)


print(
    "\nRANDOM FOREST FEATURE IMPORTANCE:"
)


display(
    rf_importance
)


# ============================================================
# 41. RANDOM FOREST FEATURE IMPORTANCE GRAPH
# ============================================================

top_features = rf_importance.head(
    10
)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    top_features["Feature"],

    top_features["Importance"]
)


plt.xlabel(
    "Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 10 Random Forest Feature Importances"
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


# ============================================================
# 42. VISUALIZE DECISION TREE
# ============================================================

plt.figure(
    figsize=(20, 10)
)


plot_tree(

    decision_tree_model,

    feature_names=X.columns,

    class_names=[
        "Low Quality",
        "High Quality"
    ],

    filled=True,

    rounded=True,

    fontsize=8
)


plt.title(
    "Wine Quality Decision Tree"
)


plt.tight_layout()

plt.show()


# ============================================================
# 43. CREATE PREDICTION RESULTS
# ============================================================

results = pd.DataFrame(

    {

        "Actual": y_test.values,

        "Decision_Tree_Predicted": dt_pred,

        "Random_Forest_Predicted": rf_pred
    }
)


# ------------------------------------------------------------
# ACTUAL LABEL
# ------------------------------------------------------------

results["Actual_Label"] = np.where(

    results["Actual"] == 1,

    "High Quality",

    "Low Quality"
)


# ------------------------------------------------------------
# DECISION TREE LABEL
# ------------------------------------------------------------

results["Decision_Tree_Label"] = np.where(

    results["Decision_Tree_Predicted"] == 1,

    "High Quality",

    "Low Quality"
)


# ------------------------------------------------------------
# RANDOM FOREST LABEL
# ------------------------------------------------------------

results["Random_Forest_Label"] = np.where(

    results["Random_Forest_Predicted"] == 1,

    "High Quality",

    "Low Quality"
)


print(
    "\nPREDICTION RESULTS:"
)


display(
    results.head(10)
)


# ============================================================
# 44. SAVE PREDICTION RESULTS
# ============================================================

results_file = (
    "wine_quality_predictions.csv"
)


results.to_csv(

    results_file,

    index=False
)


print(
    "\nPREDICTION RESULTS FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 45. CREATE FEATURE IMPORTANCE FILE
# ============================================================

feature_importance_results = pd.DataFrame(

    {

        "Feature": X.columns,

        "Decision_Tree_Importance":
        decision_tree_model.feature_importances_,

        "Random_Forest_Importance":
        random_forest_model.feature_importances_
    }
)


feature_importance_results = feature_importance_results.sort_values(

    by="Random_Forest_Importance",

    ascending=False
)


importance_file = (
    "wine_quality_feature_importance.csv"
)


feature_importance_results.to_csv(

    importance_file,

    index=False
)


print(
    "\nFEATURE IMPORTANCE FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 46. SAVE MODEL COMPARISON
# ============================================================

comparison_file = (
    "wine_quality_model_comparison.csv"
)


comparison.to_csv(

    comparison_file,

    index=False
)


print(
    "\nMODEL COMPARISON FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 47. VERIFY OUTPUT FILES
# ============================================================

print("\n" + "=" * 70)
print("PROJECT 11 COMPLETED SUCCESSFULLY!")
print("WINE QUALITY - DECISION TREE AND RANDOM FOREST")
print("=" * 70)


print("\nINPUT FILE:")
print(
    os.path.abspath(input_file)
)


print("\nTARGET COLUMN:")
print(
    target_column
)


print("\nDECISION TREE ACCURACY:")
print(
    dt_accuracy
)


print("\nRANDOM FOREST ACCURACY:")
print(
    rf_accuracy
)


print("\nOUTPUT FILE 1:")
print(
    os.path.abspath(results_file)
)


print(
    "Exists:",
    os.path.exists(results_file)
)


print("\nOUTPUT FILE 2:")
print(
    os.path.abspath(importance_file)
)


print(
    "Exists:",
    os.path.exists(importance_file)
)


print("\nOUTPUT FILE 3:")
print(
    os.path.abspath(comparison_file)
)


print(
    "Exists:",
    os.path.exists(comparison_file)
)


print("\n" + "=" * 70)
print("ALL OUTPUT FILES CREATED SUCCESSFULLY!")
print("=" * 70)
```
