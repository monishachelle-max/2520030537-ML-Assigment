```python
# ============================================================
# PROJECT 10: HEART DISEASE
# XGBOOST, LIGHTGBM AND SHAP INTERPRETABILITY
#
# Outputs:
# heart_disease_xgboost_lightgbm_results.csv
# heart_disease_feature_importance.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import glob
import warnings

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

import shap


warnings.filterwarnings("ignore")


# ============================================================
# 2. PROJECT INFORMATION
# ============================================================

print("=" * 70)
print("PROJECT 10: HEART DISEASE")
print("XGBOOST, LIGHTGBM AND SHAP INTERPRETABILITY")
print("=" * 70)


# ============================================================
# 3. CURRENT WORKING DIRECTORY
# ============================================================

print("\nCURRENT WORKING DIRECTORY:")
print(os.getcwd())


# ============================================================
# 4. FIND ALL CSV FILES
# ============================================================

csv_files = glob.glob("*.csv")


print("\nCSV FILES FOUND:")

for file in csv_files:
    print("-", file)


if len(csv_files) == 0:

    raise FileNotFoundError(
        "\nERROR: No CSV files found in this folder.\n\n"
        "Please place your Heart Disease dataset CSV file "
        "in the same folder as this notebook."
    )


# ============================================================
# 5. DEFINE POSSIBLE TARGET COLUMN NAMES
# ============================================================

possible_targets = [

    "target",
    "output",

    "heartdisease",
    "heart_disease",

    "heart_disease_present",
    "heart_disease_presence",

    "num",

    "diagnosis",
    "disease",

    "condition",

    "label",
    "class"
]


# ============================================================
# 6. AUTOMATICALLY FIND HEART DISEASE DATASET
# ============================================================

input_file = None


# ------------------------------------------------------------
# PRIORITY 1: COMMON HEART DATASET FILE NAMES
# ------------------------------------------------------------

preferred_files = [

    "heart.csv",

    "heart_disease.csv",

    "heart_disease_data.csv",

    "heart_disease_uci.csv",

    "Heart.csv",

    "Heart_Disease.csv",

    "Heart_Disease_Prediction.csv",

    "heart_failure.csv"
]


for preferred_file in preferred_files:

    if preferred_file in csv_files:

        try:

            temp_df = pd.read_csv(
                preferred_file
            )


            temp_columns = (

                temp_df.columns
                .astype(str)
                .str.strip()
                .str.lower()
                .str.replace(" ", "_", regex=False)
                .str.replace("-", "_", regex=False)
            )


            normalized_columns = [

                column
                .replace("_", "")
                .replace("-", "")
                .replace(" ", "")

                for column in temp_columns
            ]


            target_found = False


            for target in possible_targets:

                normalized_target = (

                    target
                    .replace("_", "")
                    .replace("-", "")
                    .replace(" ", "")
                )


                if normalized_target in normalized_columns:

                    target_found = True

                    break


            if target_found:

                input_file = preferred_file

                break


        except Exception:

            pass


# ------------------------------------------------------------
# PRIORITY 2: SEARCH ALL CSV FILES
# ------------------------------------------------------------

if input_file is None:

    for file in csv_files:

        try:

            temp_df = pd.read_csv(
                file,
                nrows=10
            )


            temp_columns = (

                temp_df.columns
                .astype(str)
                .str.strip()
                .str.lower()
                .str.replace(" ", "_", regex=False)
                .str.replace("-", "_", regex=False)
            )


            normalized_columns = [

                column
                .replace("_", "")
                .replace("-", "")
                .replace(" ", "")

                for column in temp_columns
            ]


            target_found = False


            for target in possible_targets:

                normalized_target = (

                    target
                    .replace("_", "")
                    .replace("-", "")
                    .replace(" ", "")
                )


                if normalized_target in normalized_columns:

                    target_found = True

                    break


            # Common heart-related features

            heart_features = [

                "age",

                "sex",

                "cp",

                "chol",

                "cholesterol",

                "trestbps",

                "restingbp",

                "resting_bp",

                "thalach",

                "maxhr",

                "exerciseangina",

                "oldpeak",

                "fbs"
            ]


            heart_feature_count = 0


            for feature in heart_features:

                normalized_feature = (

                    feature
                    .replace("_", "")
                    .replace("-", "")
                    .replace(" ", "")
                )


                if normalized_feature in normalized_columns:

                    heart_feature_count += 1


            # Select only if it looks like a Heart Disease dataset

            if target_found and heart_feature_count >= 2:

                input_file = file

                break


        except Exception:

            pass


# ============================================================
# 7. STOP IF HEART DATASET NOT FOUND
# ============================================================

if input_file is None:

    raise ValueError(
        "\nERROR: A Heart Disease dataset could not be identified.\n\n"
        "CSV files found:\n"
        + "\n".join(csv_files)
        +
        "\n\nPlease put a Heart Disease dataset CSV in this folder."
    )


print("\n" + "=" * 70)
print("HEART DISEASE DATASET FOUND SUCCESSFULLY!")
print("=" * 70)

print("\nSELECTED INPUT FILE:")
print(input_file)


# ============================================================
# 8. LOAD DATASET
# ============================================================

df = pd.read_csv(
    input_file
)


print("\nDATASET LOADED SUCCESSFULLY!")


# ============================================================
# 9. CLEAN COLUMN NAMES
# ============================================================

df.columns = (

    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)


print("\nDATASET COLUMNS:")

print(
    df.columns.tolist()
)


# ============================================================
# 10. DISPLAY DATASET
# ============================================================

print("\nFIRST 5 ROWS:")

display(
    df.head()
)


# ============================================================
# 11. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")

print(
    df.shape
)


# ============================================================
# 12. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 13. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe(
        include="all"
    )
)


# ============================================================
# 14. REPLACE MISSING VALUE SYMBOLS
# ============================================================

df = df.replace(

    [
        "?",
        " ?",
        "",
        " ",
        "NA",
        "N/A",
        "na",
        "null",
        "NULL",
        "None"
    ],

    np.nan
)


print("\nMissing value symbols replaced.")


# ============================================================
# 15. CONVERT NUMERIC-LIKE COLUMNS
# ============================================================

for column in df.columns:

    converted = pd.to_numeric(

        df[column],

        errors="coerce"
    )


    original_non_null = (

        df[column]
        .notna()
        .sum()
    )


    converted_non_null = (

        converted
        .notna()
        .sum()
    )


    # Convert only if all non-null values are numeric

    if (

        original_non_null > 0

        and

        converted_non_null == original_non_null
    ):

        df[column] = converted


# ============================================================
# 16. DISPLAY MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(
    df.isnull().sum()
)


# ============================================================
# 17. IDENTIFY TARGET COLUMN
# ============================================================

target_column = None


# ------------------------------------------------------------
# EXACT MATCH
# ------------------------------------------------------------

for target in possible_targets:

    if target in df.columns:

        target_column = target

        break


# ------------------------------------------------------------
# NORMALIZED MATCH
# ------------------------------------------------------------

if target_column is None:

    for column in df.columns:

        normalized_column = (

            column
            .lower()
            .replace("_", "")
            .replace("-", "")
            .replace(" ", "")
        )


        for target in possible_targets:

            normalized_target = (

                target
                .lower()
                .replace("_", "")
                .replace("-", "")
                .replace(" ", "")
            )


            if normalized_column == normalized_target:

                target_column = column

                break


        if target_column is not None:

            break


# ------------------------------------------------------------
# TARGET ERROR
# ------------------------------------------------------------

if target_column is None:

    print("\nAVAILABLE COLUMNS:")

    for i, column in enumerate(
        df.columns
    ):

        print(
            f"{i + 1}. {column}"
        )


    raise ValueError(
        "\nERROR: Target column could not be detected.\n"
        "Please use a valid Heart Disease dataset."
    )


print("\n" + "=" * 70)
print("TARGET COLUMN FOUND SUCCESSFULLY!")
print("=" * 70)

print("\nTARGET COLUMN:")
print(
    target_column
)


# ============================================================
# 18. PREPARE TARGET
# ============================================================

y = df[
    target_column
].copy()


# Remove missing target rows

valid_target = y.notna()


df = df.loc[
    valid_target
].copy()


y = y.loc[
    valid_target
].copy()


# ============================================================
# 19. CONVERT TARGET TO BINARY
# ============================================================

y_numeric = pd.to_numeric(

    y,

    errors="coerce"
)


# ------------------------------------------------------------
# NUMERIC TARGET
# ------------------------------------------------------------

if y_numeric.notna().sum() == len(y):

    y = y_numeric


    unique_values = sorted(
        y.unique()
    )


    print("\nORIGINAL TARGET VALUES:")

    print(
        unique_values
    )


    # Already binary

    if set(unique_values).issubset(
        {0, 1}
    ):

        y = y.astype(
            int
        )


    # UCI-style target
    # 0 = No Disease
    # >0 = Disease

    else:

        y = (

            y > 0

        ).astype(
            int
        )


# ------------------------------------------------------------
# TEXT TARGET
# ------------------------------------------------------------

else:

    y_text = (

        y.astype(str)

        .str.strip()

        .str.lower()
    )


    print("\nORIGINAL TEXT TARGET VALUES:")

    print(
        y_text.value_counts()
    )


    disease_values = [

        "1",

        "yes",

        "y",

        "disease",

        "heart disease",

        "presence",

        "present",

        "true",

        "positive",

        "abnormal"
    ]


    no_disease_values = [

        "0",

        "no",

        "n",

        "no disease",

        "absence",

        "absent",

        "false",

        "negative",

        "normal",

        "healthy"
    ]


    # Check for unknown values

    known_values = (

        disease_values

        +

        no_disease_values
    )


    unknown_values = set(

        y_text.unique()

    ) - set(
        known_values
    )


    if len(unknown_values) > 0:

        print(
            "\nWARNING: Unknown target values found:"
        )

        print(
            unknown_values
        )


    y = np.where(

        y_text.isin(
            disease_values
        ),

        1,

        0
    )


    y = pd.Series(

        y,

        index=df.index
    )


y = y.astype(
    int
)


# ============================================================
# 20. TARGET DISTRIBUTION
# ============================================================

print("\nTARGET DISTRIBUTION:")

print(
    y.value_counts()
)


if y.nunique() < 2:

    raise ValueError(
        "\nERROR: Only one target class exists.\n"
        "Please check your Heart Disease dataset."
    )


# ============================================================
# 21. TARGET DISTRIBUTION GRAPH
# ============================================================

plt.figure(
    figsize=(7, 5)
)


sns.countplot(
    x=y
)


plt.title(
    "Heart Disease Class Distribution"
)


plt.xlabel(
    "Heart Disease"
)


plt.ylabel(
    "Count"
)


plt.xticks(

    [0, 1],

    [
        "No Disease",
        "Disease"
    ]
)


plt.tight_layout()

plt.show()


# ============================================================
# 22. DEFINE FEATURES
# ============================================================

X = df.drop(

    columns=[
        target_column
    ]
)


print("\nFEATURE COLUMNS:")

print(
    X.columns.tolist()
)


# ============================================================
# 23. ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(

    X,

    drop_first=True,

    dtype=float
)


print("\nFEATURES AFTER ENCODING:")

display(
    X.head()
)


# ============================================================
# 24. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


# ============================================================
# 25. HANDLE MISSING VALUES
# ============================================================

imputer = SimpleImputer(

    strategy="median"
)


X_train_imputed = imputer.fit_transform(
    X_train
)


X_test_imputed = imputer.transform(
    X_test
)


X_train = pd.DataFrame(

    X_train_imputed,

    columns=X.columns,

    index=X_train.index
)


X_test = pd.DataFrame(

    X_test_imputed,

    columns=X.columns,

    index=X_test.index
)


print("\nMissing values handled successfully!")


print("\nTRAINING DATA SHAPE:")

print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")

print(
    X_test.shape
)


# ============================================================
# 26. CREATE XGBOOST MODEL
# ============================================================

xgb_model = XGBClassifier(

    n_estimators=200,

    learning_rate=0.05,

    max_depth=4,

    subsample=0.8,

    colsample_bytree=0.8,

    random_state=42,

    eval_metric="logloss",

    n_jobs=-1
)


# ============================================================
# 27. TRAIN XGBOOST
# ============================================================

print(
    "\nTraining XGBoost..."
)


xgb_model.fit(

    X_train,

    y_train
)


print(
    "XGBoost training completed successfully!"
)


# ============================================================
# 28. XGBOOST PREDICTIONS
# ============================================================

xgb_pred = xgb_model.predict(
    X_test
)


# ============================================================
# 29. XGBOOST EVALUATION
# ============================================================

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)


xgb_precision = precision_score(
    y_test,
    xgb_pred,
    zero_division=0
)


xgb_recall = recall_score(
    y_test,
    xgb_pred,
    zero_division=0
)


xgb_f1 = f1_score(
    y_test,
    xgb_pred,
    zero_division=0
)


print("\n" + "=" * 60)
print("XGBOOST MODEL EVALUATION")
print("=" * 60)

print(
    "Accuracy :",
    xgb_accuracy
)

print(
    "Precision:",
    xgb_precision
)

print(
    "Recall   :",
    xgb_recall
)

print(
    "F1 Score :",
    xgb_f1
)


print("\nXGBOOST CLASSIFICATION REPORT:")

print(

    classification_report(

        y_test,

        xgb_pred,

        labels=[0, 1],

        target_names=[
            "No Disease",
            "Disease"
        ],

        zero_division=0
    )
)


# ============================================================
# 30. CREATE LIGHTGBM MODEL
# ============================================================

lgb_model = LGBMClassifier(

    n_estimators=200,

    learning_rate=0.05,

    max_depth=4,

    random_state=42,

    verbosity=-1,

    n_jobs=-1
)


# ============================================================
# 31. TRAIN LIGHTGBM
# ============================================================

print(
    "\nTraining LightGBM..."
)


lgb_model.fit(

    X_train,

    y_train
)


print(
    "LightGBM training completed successfully!"
)


# ============================================================
# 32. LIGHTGBM PREDICTIONS
# ============================================================

lgb_pred = lgb_model.predict(
    X_test
)


# ============================================================
# 33. LIGHTGBM EVALUATION
# ============================================================

lgb_accuracy = accuracy_score(
    y_test,
    lgb_pred
)


lgb_precision = precision_score(
    y_test,
    lgb_pred,
    zero_division=0
)


lgb_recall = recall_score(
    y_test,
    lgb_pred,
    zero_division=0
)


lgb_f1 = f1_score(
    y_test,
    lgb_pred,
    zero_division=0
)


print("\n" + "=" * 60)
print("LIGHTGBM MODEL EVALUATION")
print("=" * 60)

print(
    "Accuracy :",
    lgb_accuracy
)

print(
    "Precision:",
    lgb_precision
)

print(
    "Recall   :",
    lgb_recall
)

print(
    "F1 Score :",
    lgb_f1
)


print("\nLIGHTGBM CLASSIFICATION REPORT:")

print(

    classification_report(

        y_test,

        lgb_pred,

        labels=[0, 1],

        target_names=[
            "No Disease",
            "Disease"
        ],

        zero_division=0
    )
)


# ============================================================
# 34. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame(

    {

        "Model": [

            "XGBoost",

            "LightGBM"
        ],

        "Accuracy": [

            xgb_accuracy,

            lgb_accuracy
        ],

        "Precision": [

            xgb_precision,

            lgb_precision
        ],

        "Recall": [

            xgb_recall,

            lgb_recall
        ],

        "F1 Score": [

            xgb_f1,

            lgb_f1
        ]
    }
)


print("\nMODEL COMPARISON:")

display(
    comparison
)


# ============================================================
# 35. MODEL COMPARISON GRAPH
# ============================================================

comparison_plot = comparison.set_index(
    "Model"
)


comparison_plot.plot(

    kind="bar",

    figsize=(10, 6)
)


plt.title(
    "XGBoost vs LightGBM Performance"
)


plt.ylabel(
    "Score"
)


plt.ylim(
    0,
    1
)


plt.tight_layout()

plt.show()


# ============================================================
# 36. XGBOOST CONFUSION MATRIX
# ============================================================

xgb_cm = confusion_matrix(

    y_test,

    xgb_pred,

    labels=[0, 1]
)


plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    xgb_cm,

    annot=True,

    fmt="d",

    xticklabels=[
        "No Disease",
        "Disease"
    ],

    yticklabels=[
        "No Disease",
        "Disease"
    ]
)


plt.title(
    "XGBoost Confusion Matrix"
)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.tight_layout()

plt.show()


# ============================================================
# 37. LIGHTGBM CONFUSION MATRIX
# ============================================================

lgb_cm = confusion_matrix(

    y_test,

    lgb_pred,

    labels=[0, 1]
)


plt.figure(
    figsize=(7, 5)
)


sns.heatmap(

    lgb_cm,

    annot=True,

    fmt="d",

    xticklabels=[
        "No Disease",
        "Disease"
    ],

    yticklabels=[
        "No Disease",
        "Disease"
    ]
)


plt.title(
    "LightGBM Confusion Matrix"
)


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.tight_layout()

plt.show()


# ============================================================
# 38. XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": xgb_model.feature_importances_
    }
)


xgb_importance = xgb_importance.sort_values(

    by="Importance",

    ascending=False
)


print(
    "\nXGBOOST FEATURE IMPORTANCE:"
)


display(
    xgb_importance
)


# ============================================================
# 39. LIGHTGBM FEATURE IMPORTANCE
# ============================================================

lgb_importance = pd.DataFrame(

    {

        "Feature": X.columns,

        "Importance": lgb_model.feature_importances_
    }
)


lgb_importance = lgb_importance.sort_values(

    by="Importance",

    ascending=False
)


print(
    "\nLIGHTGBM FEATURE IMPORTANCE:"
)


display(
    lgb_importance
)


# ============================================================
# 40. XGBOOST FEATURE IMPORTANCE GRAPH
# ============================================================

top_xgb_features = xgb_importance.head(
    10
)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    top_xgb_features["Feature"],

    top_xgb_features["Importance"]
)


plt.xlabel(
    "Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 10 XGBoost Feature Importances"
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


# ============================================================
# 41. LIGHTGBM FEATURE IMPORTANCE GRAPH
# ============================================================

top_lgb_features = lgb_importance.head(
    10
)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    top_lgb_features["Feature"],

    top_lgb_features["Importance"]
)


plt.xlabel(
    "Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 10 LightGBM Feature Importances"
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


# ============================================================
# 42. SHAP INTERPRETABILITY
# ============================================================

print("\n" + "=" * 60)
print("SHAP INTERPRETABILITY - XGBOOST")
print("=" * 60)


try:

    explainer = shap.TreeExplainer(
        xgb_model
    )


    shap_values = explainer(
        X_test
    )


    print(
        "\nSHAP values calculated successfully!"
    )


    # ========================================================
    # SHAP BAR SUMMARY PLOT
    # ========================================================

    print(
        "\nDisplaying SHAP Feature Importance Plot..."
    )


    shap.summary_plot(

        shap_values.values,

        X_test,

        plot_type="bar",

        show=True
    )


    # ========================================================
    # SHAP DETAILED SUMMARY PLOT
    # ========================================================

    print(
        "\nDisplaying Detailed SHAP Summary Plot..."
    )


    shap.summary_plot(

        shap_values.values,

        X_test,

        show=True
    )


except Exception as shap_error:

    print(
        "\nSHAP could not create plots."
    )

    print(
        "Reason:",
        shap_error
    )


# ============================================================
# 43. CREATE PREDICTION RESULTS
# ============================================================

results = pd.DataFrame(

    {

        "Actual": y_test.values,

        "XGBoost_Predicted": xgb_pred,

        "LightGBM_Predicted": lgb_pred
    }
)


results["Actual_Label"] = np.where(

    results["Actual"] == 1,

    "Disease",

    "No Disease"
)


results["XGBoost_Label"] = np.where(

    results["XGBoost_Predicted"] == 1,

    "Disease",

    "No Disease"
)


results["LightGBM_Label"] = np.where(

    results["LightGBM_Predicted"] == 1,

    "Disease",

    "No Disease"
)


print(
    "\nPREDICTION RESULTS:"
)


display(
    results.head(10)
)


# ============================================================
# 44. SAVE PREDICTION RESULTS
# ============================================================

results_file = (
    "heart_disease_xgboost_lightgbm_results.csv"
)


results.to_csv(

    results_file,

    index=False
)


print(
    "\nRESULTS FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 45. SAVE FEATURE IMPORTANCE
# ============================================================

feature_importance_results = pd.DataFrame(

    {

        "Feature": X.columns,

        "XGBoost_Importance":
        xgb_model.feature_importances_,

        "LightGBM_Importance":
        lgb_model.feature_importances_
    }
)


feature_importance_results = (

    feature_importance_results

    .sort_values(

        by="XGBoost_Importance",

        ascending=False
    )
)


importance_file = (
    "heart_disease_feature_importance.csv"
)


feature_importance_results.to_csv(

    importance_file,

    index=False
)


print(
    "\nFEATURE IMPORTANCE FILE SAVED SUCCESSFULLY!"
)


# ============================================================
# 46. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PROJECT 10 COMPLETED SUCCESSFULLY!")
print("HEART DISEASE - XGBOOST, LIGHTGBM AND SHAP")
print("=" * 70)


print("\nINPUT FILE:")

print(
    os.path.abspath(
        input_file
    )
)


print("\nTARGET COLUMN:")

print(
    target_column
)


print("\nXGBOOST ACCURACY:")

print(
    xgb_accuracy
)


print("\nLIGHTGBM ACCURACY:")

print(
    lgb_accuracy
)


print("\nRESULTS FILE:")

print(
    os.path.abspath(
        results_file
    )
)


print(
    "Exists:",
    os.path.exists(
        results_file
    )
)


print("\nFEATURE IMPORTANCE FILE:")

print(
    os.path.abspath(
        importance_file
    )
)


print(
    "Exists:",
    os.path.exists(
        importance_file
    )
)


print("\n" + "=" * 70)
print("ALL OUTPUT FILES CREATED SUCCESSFULLY!")
print("=" * 70)
```
