```python
# ============================================================
# PROJECT 4: CALIFORNIA HOUSING
# LINEAR REGRESSION AND REGULARISATION
#
# Output:
# california_housing_model_results.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    Lasso
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. PROJECT INFORMATION
# ============================================================

print("=" * 65)
print("PROJECT 4: CALIFORNIA HOUSING")
print("LINEAR REGRESSION AND REGULARISATION")
print("=" * 65)


# ============================================================
# 3. CHECK CURRENT WORKING DIRECTORY
# ============================================================

print("\nCURRENT WORKING DIRECTORY:")
print(os.getcwd())


# ============================================================
# 4. LOAD CALIFORNIA HOUSING DATASET
# ============================================================

print("\nLOADING CALIFORNIA HOUSING DATASET...")


housing = fetch_california_housing(
    as_frame=True
)


df = housing.frame.copy()


print("\nDATASET LOADED SUCCESSFULLY!")


# ============================================================
# 5. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")

display(
    df.head()
)


# ============================================================
# 6. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")

print(
    df.shape
)


# ============================================================
# 7. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")

df.info()


# ============================================================
# 8. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")

display(
    df.describe()
)


# ============================================================
# 9. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES:")

print(
    df.isnull().sum()
)


# ============================================================
# 10. TARGET COLUMN
# ============================================================

target_column = "MedHouseVal"


print("\nTARGET COLUMN:")

print(
    target_column
)


# ============================================================
# 11. TARGET DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(8, 5)
)


sns.histplot(
    df[target_column],
    kde=True
)


plt.title(
    "California Housing Target Distribution"
)


plt.xlabel(
    "Median House Value"
)


plt.ylabel(
    "Count"
)


plt.tight_layout()

plt.show()


# ============================================================
# 12. CORRELATION HEATMAP
# ============================================================

plt.figure(
    figsize=(12, 8)
)


sns.heatmap(
    df.corr(),
    annot=True,
    fmt=".2f"
)


plt.title(
    "Feature Correlation Heatmap"
)


plt.tight_layout()

plt.show()


# ============================================================
# 13. DEFINE FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=[
        target_column
    ]
)


y = df[
    target_column
]


print("\nFEATURE COLUMNS:")

print(
    X.columns.tolist()
)


# ============================================================
# 14. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42
)


print("\nTRAINING DATA SHAPE:")

print(
    X_train.shape
)


print("\nTESTING DATA SHAPE:")

print(
    X_test.shape
)


# ============================================================
# 15. FEATURE SCALING
# ============================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


print(
    "\nFeature scaling completed successfully!"
)


# ============================================================
# 16. CREATE LINEAR REGRESSION MODEL
# ============================================================

linear_model = LinearRegression()


print(
    "\nLinear Regression Model Created!"
)


# ============================================================
# 17. TRAIN LINEAR REGRESSION
# ============================================================

linear_model.fit(

    X_train_scaled,

    y_train
)


print(
    "Linear Regression Model Trained Successfully!"
)


# ============================================================
# 18. LINEAR REGRESSION PREDICTIONS
# ============================================================

linear_pred = linear_model.predict(
    X_test_scaled
)


# ============================================================
# 19. CREATE RIDGE REGRESSION MODEL
# ============================================================

ridge_model = Ridge(

    alpha=1.0
)


print(
    "\nRidge Regression Model Created!"
)


# ============================================================
# 20. TRAIN RIDGE REGRESSION
# ============================================================

ridge_model.fit(

    X_train_scaled,

    y_train
)


print(
    "Ridge Regression Model Trained Successfully!"
)


# ============================================================
# 21. RIDGE REGRESSION PREDICTIONS
# ============================================================

ridge_pred = ridge_model.predict(
    X_test_scaled
)


# ============================================================
# 22. CREATE LASSO REGRESSION MODEL
# ============================================================

lasso_model = Lasso(

    alpha=0.01,

    max_iter=10000
)


print(
    "\nLasso Regression Model Created!"
)


# ============================================================
# 23. TRAIN LASSO REGRESSION
# ============================================================

lasso_model.fit(

    X_train_scaled,

    y_train
)


print(
    "Lasso Regression Model Trained Successfully!"
)


# ============================================================
# 24. LASSO REGRESSION PREDICTIONS
# ============================================================

lasso_pred = lasso_model.predict(
    X_test_scaled
)


# ============================================================
# 25. MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model_name,
    y_actual,
    y_predicted
):

    mae = mean_absolute_error(
        y_actual,
        y_predicted
    )


    mse = mean_squared_error(
        y_actual,
        y_predicted
    )


    rmse = np.sqrt(
        mse
    )


    r2 = r2_score(
        y_actual,
        y_predicted
    )


    return {

        "Model": model_name,

        "MAE": mae,

        "MSE": mse,

        "RMSE": rmse,

        "R2 Score": r2
    }


# ============================================================
# 26. EVALUATE ALL MODELS
# ============================================================

linear_results = evaluate_model(

    "Linear Regression",

    y_test,

    linear_pred
)


ridge_results = evaluate_model(

    "Ridge Regression",

    y_test,

    ridge_pred
)


lasso_results = evaluate_model(

    "Lasso Regression",

    y_test,

    lasso_pred
)


# ============================================================
# 27. CREATE MODEL COMPARISON TABLE
# ============================================================

comparison = pd.DataFrame(

    [

        linear_results,

        ridge_results,

        lasso_results
    ]
)


comparison = comparison.sort_values(

    by="R2 Score",

    ascending=False
)


print("\n" + "=" * 65)

print("MODEL COMPARISON")

print("=" * 65)


display(
    comparison
)


# ============================================================
# 28. MODEL PERFORMANCE GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.bar(

    comparison["Model"],

    comparison["R2 Score"]
)


plt.title(
    "Model Comparison - R2 Score"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "R2 Score"
)


plt.ylim(
    0,
    1
)


plt.tight_layout()

plt.show()


# ============================================================
# 29. RMSE COMPARISON GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.bar(

    comparison["Model"],

    comparison["RMSE"]
)


plt.title(
    "Model Comparison - RMSE"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "RMSE"
)


plt.tight_layout()

plt.show()


# ============================================================
# 30. ACTUAL VS PREDICTED - LINEAR REGRESSION
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.scatter(

    y_test,

    linear_pred
)


plt.xlabel(
    "Actual House Value"
)


plt.ylabel(
    "Predicted House Value"
)


plt.title(
    "Linear Regression - Actual vs Predicted"
)


min_value = min(

    y_test.min(),

    linear_pred.min()
)


max_value = max(

    y_test.max(),

    linear_pred.max()
)


plt.plot(

    [

        min_value,

        max_value
    ],

    [

        min_value,

        max_value
    ]
)


plt.tight_layout()

plt.show()


# ============================================================
# 31. FEATURE COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame(

    {

        "Feature": X.columns,

        "Linear_Coefficient":
            linear_model.coef_,

        "Ridge_Coefficient":
            ridge_model.coef_,

        "Lasso_Coefficient":
            lasso_model.coef_
    }
)


print(
    "\nFEATURE COEFFICIENTS:"
)


display(
    coefficients
)


# ============================================================
# 32. CREATE PREDICTION RESULTS DATASET
# ============================================================

results = pd.DataFrame(

    {

        "Actual": y_test.values,

        "Linear_Regression_Predicted":
            linear_pred,

        "Ridge_Regression_Predicted":
            ridge_pred,

        "Lasso_Regression_Predicted":
            lasso_pred
    }
)


print(
    "\nPREDICTION RESULTS:"
)


display(
    results.head(10)
)


# ============================================================
# 33. SAVE MODEL COMPARISON RESULTS
# ============================================================

output_file = (
    "california_housing_model_results.csv"
)


comparison.to_csv(

    output_file,

    index=False
)


print(
    "\nMODEL RESULTS SAVED SUCCESSFULLY!"
)


# ============================================================
# 34. VERIFY OUTPUT FILE
# ============================================================

print("\n" + "=" * 65)

print("PROJECT 4 COMPLETED SUCCESSFULLY!")

print(
    "California Housing — Linear Regression and Regularisation"
)

print("=" * 65)


print("\nModels Used:")

print(
    "1. Linear Regression"
)

print(
    "2. Ridge Regression"
)

print(
    "3. Lasso Regression"
)


print("\nOutput File:")

print(
    output_file
)


print("\nFile Created:")

print(
    os.path.exists(
        output_file
    )
)


print("\nExact File Location:")

print(
    os.path.abspath(
        output_file
    )
)


print("\n" + "=" * 65)
```
