```python

# ============================================================
# PROJECT 3: ADULT INCOME — FEATURE ENGINEERING AND EDA
# File Name: 03_Adult_Income_EDA.ipynb
# Input Dataset: adult.csv
# Output Dataset: adult_feature_engineered.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 2. DISPLAY SETTINGS
# ============================================================

pd.set_option("display.max_columns", None)


# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv("adult.csv")

print("Dataset loaded successfully!")


# ============================================================
# 4. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")
display(df.head())


# ============================================================
# 5. DATASET SHAPE
# ============================================================

print("\nDATASET SHAPE:")
print(df.shape)


# ============================================================
# 6. ORIGINAL COLUMN NAMES
# ============================================================

print("\nORIGINAL COLUMN NAMES:")
print(df.columns.tolist())


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "-", regex=False)
    .str.replace("_", "-", regex=False)
)


print("\nCLEANED COLUMN NAMES:")
print(df.columns.tolist())


# ============================================================
# 8. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")
df.info()


# ============================================================
# 9. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY:")
display(df.describe(include="all"))


# ============================================================
# 10. REPLACE ? WITH NaN
# ============================================================

df = df.replace(
    [
        "?",
        " ?",
        "? ",
        ""
    ],
    np.nan
)


# ============================================================
# 11. REMOVE EXTRA SPACES FROM TEXT COLUMNS
# ============================================================

object_columns = df.select_dtypes(
    include="object"
).columns


for column in object_columns:

    df[column] = df[column].str.strip()


# ============================================================
# 12. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES BEFORE HANDLING:")
print(df.isnull().sum())


# ============================================================
# 13. HANDLE MISSING VALUES
# ============================================================

for column in df.columns:

    if df[column].dtype == "object":

        if df[column].isnull().sum() > 0:

            df[column] = df[column].fillna(
                df[column].mode()[0]
            )

    else:

        if df[column].isnull().sum() > 0:

            df[column] = df[column].fillna(
                df[column].median()
            )


print("\nMISSING VALUES AFTER HANDLING:")
print(df.isnull().sum())


# ============================================================
# 14. FIND IMPORTANT COLUMNS
# ============================================================

age_column = None
income_column = None
education_column = None
gender_column = None
capital_gain_column = None
hours_column = None


# Find Age Column
if "age" in df.columns:
    age_column = "age"


# Find Income Column
possible_income_columns = [
    "income",
    "income-level",
    "salary"
]


for column in possible_income_columns:

    if column in df.columns:

        income_column = column

        break


# Find Education Column
if "education" in df.columns:
    education_column = "education"


# Find Gender / Sex Column
possible_gender_columns = [
    "gender",
    "sex"
]


for column in possible_gender_columns:

    if column in df.columns:

        gender_column = column

        break


# Find Capital Gain Column
possible_capital_gain_columns = [
    "capital-gain",
    "capitalgain"
]


for column in possible_capital_gain_columns:

    if column in df.columns:

        capital_gain_column = column

        break


# Find Hours Per Week Column
possible_hours_columns = [
    "hours-per-week",
    "hoursperweek"
]


for column in possible_hours_columns:

    if column in df.columns:

        hours_column = column

        break


print("\nIMPORTANT COLUMNS DETECTED:")

print("Age Column:", age_column)

print("Income Column:", income_column)

print("Education Column:", education_column)

print("Gender Column:", gender_column)

print("Capital Gain Column:", capital_gain_column)

print("Hours Per Week Column:", hours_column)


# ============================================================
# 15. EDA — INCOME DISTRIBUTION
# ============================================================

if income_column is not None:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x=income_column
    )

    plt.title("Income Distribution")

    plt.xlabel("Income")

    plt.ylabel("Count")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 16. EDA — AGE DISTRIBUTION
# ============================================================

if age_column is not None:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df,
        x=age_column,
        kde=True,
        bins=30
    )

    plt.title("Age Distribution")

    plt.xlabel("Age")

    plt.ylabel("Count")

    plt.tight_layout()

    plt.show()


# ============================================================
# 17. EDA — EDUCATION VS INCOME
# ============================================================

if (
    education_column is not None
    and income_column is not None
):

    plt.figure(figsize=(14, 6))

    sns.countplot(
        data=df,
        x=education_column,
        hue=income_column
    )

    plt.title("Education vs Income")

    plt.xlabel("Education")

    plt.ylabel("Count")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# 18. EDA — GENDER VS INCOME
# ============================================================

if (
    gender_column is not None
    and income_column is not None
):

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x=gender_column,
        hue=income_column
    )

    plt.title("Gender vs Income")

    plt.xlabel("Gender")

    plt.ylabel("Count")

    plt.tight_layout()

    plt.show()


# ============================================================
# 19. FEATURE ENGINEERING — AGE GROUP
# ============================================================

if age_column is not None:

    df["age-group"] = pd.cut(
        df[age_column],
        bins=[
            0,
            25,
            40,
            60,
            100
        ],
        labels=[
            "Young",
            "Adult",
            "Middle_Age",
            "Senior"
        ],
        include_lowest=True
    )


    print("\nAGE GROUP FEATURE CREATED!")

    display(
        df[
            [
                age_column,
                "age-group"
            ]
        ].head()
    )


# ============================================================
# 20. FEATURE ENGINEERING — CAPITAL GAIN INDICATOR
# ============================================================

if capital_gain_column is not None:

    df["has-capital-gain"] = (
        df[capital_gain_column] > 0
    ).astype(int)


    print("\nCAPITAL GAIN FEATURE CREATED!")

    display(
        df[
            [
                capital_gain_column,
                "has-capital-gain"
            ]
        ].head()
    )


# ============================================================
# 21. FEATURE ENGINEERING — WORK HOURS CATEGORY
# ============================================================

if hours_column is not None:

    df["work-hours-category"] = pd.cut(
        df[hours_column],
        bins=[
            0,
            20,
            40,
            60,
            np.inf
        ],
        labels=[
            "Part_Time",
            "Normal",
            "Long_Hours",
            "Very_Long_Hours"
        ],
        include_lowest=True
    )


    print("\nWORK HOURS CATEGORY FEATURE CREATED!")

    display(
        df[
            [
                hours_column,
                "work-hours-category"
            ]
        ].head()
    )


# ============================================================
# 22. EDA — INCOME BY AGE GROUP
# ============================================================

if (
    "age-group" in df.columns
    and income_column is not None
):

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        x="age-group",
        hue=income_column
    )

    plt.title("Income by Age Group")

    plt.xlabel("Age Group")

    plt.ylabel("Count")

    plt.tight_layout()

    plt.show()


# ============================================================
# 23. EDA — INCOME BY WORK HOURS
# ============================================================

if (
    "work-hours-category" in df.columns
    and income_column is not None
):

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        x="work-hours-category",
        hue=income_column
    )

    plt.title("Income by Work Hours Category")

    plt.xlabel("Work Hours Category")

    plt.ylabel("Count")

    plt.tight_layout()

    plt.show()


# ============================================================
# 24. NUMERICAL DATA FOR CORRELATION
# ============================================================

numeric_df = df.select_dtypes(
    include=[
        "number"
    ]
)


print("\nNUMERICAL COLUMNS:")
print(
    numeric_df.columns.tolist()
)


# ============================================================
# 25. CORRELATION HEATMAP
# ============================================================

plt.figure(
    figsize=(14, 10)
)


sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f"
)


plt.title(
    "Correlation Heatmap"
)


plt.tight_layout()

plt.show()


# ============================================================
# 26. FINAL DATASET PREVIEW
# ============================================================

print("\nFINAL FEATURE ENGINEERED DATASET:")

display(
    df.head()
)


# ============================================================
# 27. FINAL DATASET SHAPE
# ============================================================

print(
    "\nFINAL DATASET SHAPE:"
)

print(
    df.shape
)


# ============================================================
# 28. FINAL MISSING VALUES CHECK
# ============================================================

print(
    "\nFINAL MISSING VALUES:"
)

print(
    df.isnull().sum()
)


# ============================================================
# 29. SAVE FEATURE ENGINEERED DATASET
# ============================================================

df.to_csv(
    "adult_feature_engineered.csv",
    index=False
)


# ============================================================
# 30. PROJECT COMPLETION MESSAGE
# ============================================================

print(
    "\n=============================================="
)

print(
    "PROJECT 3 COMPLETED SUCCESSFULLY!"
)

print(
    "Adult Income — Feature Engineering and EDA"
)

print(
    "Output saved as:"
)

print(
    "adult_feature_engineered.csv"
)

print(
    "=============================================="
)
```
