```python
# ============================================================
# PROJECT 2: TITANIC SURVIVAL
# PREPROCESSING PIPELINE AND CLEANED DATASET
#
# Notebook: 02_Titanic_Preprocessing.ipynb
# Input:    titanic.csv
# Output:   titanic_cleaned_dataset.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import os


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("titanic.csv")

print("Dataset loaded successfully!")


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFIRST 5 ROWS:")
display(df.head())


# ============================================================
# 4. DATASET SHAPE
# ============================================================

print("\nORIGINAL DATASET SHAPE:")
print(df.shape)


# ============================================================
# 5. COLUMN NAMES
# ============================================================

print("\nCOLUMN NAMES:")
print(df.columns.tolist())


# ============================================================
# 6. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
)

print("\nCLEANED COLUMN NAMES:")
print(df.columns.tolist())


# ============================================================
# 7. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION:")
df.info()


# ============================================================
# 8. CHECK DUPLICATE ROWS
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nDUPLICATE ROWS:")
print(duplicate_count)


# Remove duplicate rows if they exist

df = df.drop_duplicates()

print("\nDATASET SHAPE AFTER REMOVING DUPLICATES:")
print(df.shape)


# ============================================================
# 9. CHECK MISSING VALUES BEFORE PREPROCESSING
# ============================================================

print("\nMISSING VALUES BEFORE PREPROCESSING:")
print(df.isnull().sum())


# ============================================================
# 10. DROP UNNECESSARY COLUMNS
# ============================================================

columns_to_drop = [
    "name",
    "ticket",
    "cabin",
    "boat",
    "body",
    "home.dest"
]


existing_columns = [
    column
    for column in columns_to_drop
    if column in df.columns
]


if existing_columns:

    df = df.drop(
        columns=existing_columns
    )

    print("\nDROPPED COLUMNS:")
    print(existing_columns)

else:

    print("\nNo unnecessary columns found to drop.")


# ============================================================
# 11. HANDLE MISSING NUMERICAL VALUES
# ============================================================

numerical_columns = df.select_dtypes(
    include=["number"]
).columns


for column in numerical_columns:

    if df[column].isnull().sum() > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(
            median_value
        )

        print(
            f"Missing values filled in numerical column: {column}"
        )


# ============================================================
# 12. HANDLE MISSING CATEGORICAL VALUES
# ============================================================

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns


for column in categorical_columns:

    if df[column].isnull().sum() > 0:

        mode_value = df[column].mode()[0]

        df[column] = df[column].fillna(
            mode_value
        )

        print(
            f"Missing values filled in categorical column: {column}"
        )


# ============================================================
# 13. CHECK MISSING VALUES AFTER HANDLING
# ============================================================

print("\nMISSING VALUES AFTER HANDLING:")
print(df.isnull().sum())


# ============================================================
# 14. ENCODE SEX COLUMN
# ============================================================

if "sex" in df.columns:

    df["sex"] = (
        df["sex"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(
            {
                "male": 1,
                "female": 0
            }
        )
    )

    print("\nSEX COLUMN ENCODED:")
    print("Male = 1")
    print("Female = 0")

else:

    print("\nSex column not found.")


# ============================================================
# 15. ONE-HOT ENCODE CATEGORICAL COLUMNS
# ============================================================

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()


if categorical_columns:

    print("\nCATEGORICAL COLUMNS TO ENCODE:")
    print(categorical_columns)

    df = pd.get_dummies(
        df,
        columns=categorical_columns,
        drop_first=True,
        dtype=int
    )

    print("\nCategorical columns encoded successfully!")

else:

    print("\nNo categorical columns found for encoding.")


# ============================================================
# 16. CONVERT BOOLEAN COLUMNS TO INTEGER
# ============================================================

boolean_columns = df.select_dtypes(
    include=["bool"]
).columns


for column in boolean_columns:

    df[column] = df[column].astype(int)


# ============================================================
# 17. FINAL DATA TYPES
# ============================================================

print("\nFINAL DATA TYPES:")
print(df.dtypes)


# ============================================================
# 18. FINAL DATASET PREVIEW
# ============================================================

print("\nCLEANED DATASET PREVIEW:")
display(df.head())


# ============================================================
# 19. FINAL DATASET SHAPE
# ============================================================

print("\nFINAL DATASET SHAPE:")
print(df.shape)


# ============================================================
# 20. FINAL MISSING VALUES CHECK
# ============================================================

print("\nFINAL MISSING VALUES:")
print(df.isnull().sum())


# ============================================================
# 21. CHECK FINAL DUPLICATES
# ============================================================

print("\nFINAL DUPLICATE ROWS:")
print(df.duplicated().sum())


# ============================================================
# 22. SAVE CLEANED DATASET
# ============================================================

output_file = "titanic_cleaned_dataset.csv"

df.to_csv(
    output_file,
    index=False
)


# ============================================================
# 23. VERIFY FILE CREATION
# ============================================================

file_exists = os.path.exists(output_file)

print("\n==============================================")
print("PROJECT 2 COMPLETED SUCCESSFULLY!")
print("Titanic Survival — Preprocessing Pipeline")
print("==============================================")

print("\nOutput File:")
print(output_file)

print("\nFile Created:")
print(file_exists)

print("\nExact File Location:")
print(os.path.abspath(output_file))

print("\nFinal Dataset Shape:")
print(df.shape)

print("\n==============================================")
```
